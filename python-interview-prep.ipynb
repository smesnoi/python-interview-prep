# Задание 1. Пересечение без дубликатов

**Задача:** найти пересечение двух списков чисел без дубликатов.

**Алгоритм:**
1. Считываем две строки через `input()`.
2. Конвертируем в списки целых чисел (`split(',')` + `strip()` + `int()`).
3. Функция `pure_intersection`:
   - Преобразует списки во множества (`set`).
   - Находит пересечение через `set.intersection()`.
   - Сохраняет порядок элементов как в первом списке (убирает дубликаты).
4. Обрабатываем ошибку `ValueError` → «Некорректный ввод».

**Тесты:**
- Пример 1: `[42, 52, 14]` 
- Пример 2: `[]` 
- Пример 3: `Некорректный ввод` 

```python
def pure_intersection(list1, list2):
    """
    Возвращает список уникальных элементов, которые есть и в list1, и в list2.
    Порядок элементов — как в list1.
    
    Параметры:
        list1 (list) — первый список чисел,
        list2 (list) — второй список чисел.
    
    Возвращает:
        list — уникальные элементы пересечения в порядке из list1.
    """
    # Преобразуем оба списка во множества и находим пересечение
    intersection = set(list1).intersection(set(list2))
    
    # Сохраняем порядок элементов как в list1 (убираем дубликаты)
    result = []
    seen = set()
    for x in list1:
        if x in intersection and x not in seen:
            result.append(x)
            seen.add(x)
    
    return result


# Основной блок программы
try:
    # Ввод пользователя (две строки через input())
    input1 = input("Введите первый список: ")
    input2 = input("Введите второй список: ")
    
    # Конвертируем строки в списки целых чисел
    # split(',') — разделяем по запятой, strip() убирает пробелы
    list1 = [int(x.strip()) for x in input1.split(',')]
    list2 = [int(x.strip()) for x in input2.split(',')]
    
    # Находим пересечение
    result = pure_intersection(list1, list2)
    
    # Выводим результат
    print("Результат:", result)

except ValueError:
    # Если ввод нельзя конвертировать в целые числа — выводим ошибку
    print("Некорректный ввод")
```

# Задание 2. Минимум и максимум

**Задача:** найти min и max из последовательности чисел, введённых через пробел.

**Особенности ввода:**
- Дробная часть — **запятая** (`-2,56`), а не точка.
- Слова (`восемь`) **игнорируются**.

**Алгоритм:**
1. Считываем строку через `input()`, разбиваем по пробелам (`split()`).
2. Для каждого токена:
   - Если `token.isalpha()` → пропускаем (слово).
   - Иначе заменяем `,` на `.` и конвертируем в `float`.
   - При ошибке (`ValueError`) — пропускаем.
3. Функция `find_min_max(numbers)` возвращает `(min, max)`.
4. Выводим результат.

**Тест:**
- Ввод: `-2,56 25,002 восемь 19,13 13 -37,5`
- Вывод: `Minimum: -37.5`, `Maximum: 25.002` 

```python
def find_min_max(numbers):
    """
    Возвращает минимум и максимум из списка чисел.
    
    Параметры:
        numbers (list) — список действительных чисел.
    
    Возвращает:
        tuple (min_value, max_value).
    """
    return min(numbers), max(numbers)


# Основной блок программы
try:
    # Ввод пользователя (одна строка через input())
    user_input = input("Введите числа через пробел: ")
    
    # Разбиваем строку по пробелам на отдельные токены
    tokens = user_input.split()
    
    # Список для чисел
    numbers = []
    
    for token in tokens:
        # Если токен состоит только из букв — игнорируем (например, "восемь")
        if token.isalpha():
            continue
        
        # Заменяем запятую на точку (для корректной конвертации во float)
        # и добавляем число в список
        try:
            numbers.append(float(token.replace(',', '.')))
        except ValueError:
            # Если токен — не число и не слово (например, "5.5.5") — игнорируем
            continue
    
    # Находим минимум и максимум
    min_val, max_val = find_min_max(numbers)
    
    # Выводим результат
    print(f"Minimum: {min_val}")
    print(f"Maximum: {max_val}")

except ValueError:
    print("Некорректный ввод")
```

# Задание 3. Медиана

**Задача:** найти медиану последовательности чисел.

**Медиана** — центральный элемент упорядоченной последовательности:
- Нечётная длина → центральный элемент.
- Чётная длина → среднее двух центральных.

**Алгоритм:**
1. Считываем строку через `input()`, разбиваем по запятой.
2. Конвертируем в список `float`. Если слово → `ValueError` → «Некорректный ввод».
3. Функция `find_median(numbers)`:
   - Пустой список → `None`.
   - Сортировка → `sorted(numbers)`.
   - Нечётная длина → `sorted[n//2]`.
   - Чётная длина → `(sorted[n//2-1] + sorted[n//2]) / 2`.
4. Выводим результат.

**Тесты:**
- `1, 5, 2, 3, 6` → `Median: 3.0` 
- `100, 5, 2, 4, 3, 6` → `Median: 4.5` 
- (пусто) → `Некорректный ввод` 
- `десять, 10, пять, 7, семь` → `Некорректный ввод` 

```python
def find_median(numbers):
    """
    Возвращает медиану списка чисел.
    
    Медиана — центральный элемент упорядоченной последовательности.
    Если длина чётная — среднее между двумя центральными элементами.
    
    Параметры:
        numbers (list) — список действительных чисел.
    
    Возвращает:
        float — медиана, или None, если список пустой.
    """
    # Если список пустой — возвращаем None
    if not numbers:
        return None
    
    # Сортируем список по возрастанию
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    
    # Если длина нечётная — центральный элемент
    if n % 2 == 1:
        return float(sorted_numbers[n // 2])
    # Если длина чётная — среднее между двумя центральными
    else:
        return (sorted_numbers[n // 2 - 1] + sorted_numbers[n // 2]) / 2


# Основной блок программы
try:
    # Ввод пользователя
    user_input = input("Введите числа через запятую и пробел: ")
    
    # Если ввод пустой — ошибка
    if not user_input.strip():
        print("Некорректный ввод")
    else:
        # Разбиваем по запятой и убираем пробелы
        tokens = [t.strip() for t in user_input.split(',')]
        
        # Конвертируем в числа (если слово — ValueError → «Некорректный ввод»)
        numbers = [float(t) for t in tokens if t != '']
        
        # Считаем медиану
        median = find_median(numbers)
        
        # Выводим результат
        if median is None:
            print("Некорректный ввод")
        else:
            print(f"Median: {median}")

except ValueError:
    # Если ввод нельзя конвертировать в числа — ошибка
    print("Некорректный ввод")
```

# Задание 4. Преобразование строки в число

**Задача:** преобразовать описание числа словами в целое число (от 0 до 1 000 000).

**Алгоритм:**
1. Разбиваем строку на слова (`split()`).
2. Для каждого слова ищем **самый длинный** префикс из `number_word_dict`.
3. Если префикс — `ты` (1000): `current *= 1000`, `total += current`, `current = 0`.
4. Если префикс — `м` (1 000 000): `current *= 1000000`, `total += current`, `current = 0`.
5. Иначе: `current += значение`.
6. В конце: `total += current`.

**Почему самый длинный префикс?**
- `сто` (100) vs `с` (7): берём `сто`, потому что 3 буквы > 1 буквы.
- `двад` (20) vs `дв` (2): берём `двад`.
- Это позволяет правильно различать `семь` (7) и `семьсот` (700).

**Тесты:**
- `один` → `1` 
- `двадцать` → `20` 
- `двести сорок шесть` → `246` 
- `семьсот восемьдесят три тысячи девятьсот девятнадцать` → `783919` 

```python
def transform_string_to_integer(s):
    """
    Преобразует строку — описание числа словами — в целое число.
    
    Работает для чисел от 0 до 1 000 000 включительно.
    
    Параметры:
        s (str) — строка с описанием числа (например, "двести сорок шесть").
    
    Возвращает:
        int — число, описанное в строке.
    """
    # Словарь префиксов: ключ — начало слова, значение — число
    number_word_dict = {
        "ты": 1000, "м": 1000000,
        "сто": 100, "двес": 200, "трис": 300, "четырес": 400,
        "пятьс": 500, "шестьс": 600, "семьс": 700, "восемьс": 800, "девятьс": 900,
        "одинн": 11, "двен": 12, "трин": 13, "четырн": 14, "пятн": 15,
        "шестн": 16, "семн": 17, "восемн": 18, "девятн": 19,
        "двад": 20, "трид": 30, "сор": 40, "пятьд": 50,
        "шестьд": 60, "семьд": 70, "восемьд": 80, "девяно": 90,
        "дес": 10, "н": 0, "о": 1, "дв": 2, "т": 3, "ч": 4,
        "п": 5, "ш": 6, "с": 7, "в": 8, "д": 9,
    }
    
    # Разбиваем строку на отдельные слова
    words = s.split()
    
    # Переменные для накопления результата
    total = 0        # итоговое число
    current = 0      # текущая сумма (до тысяч/миллионов)
    
    for word in words:
        # Ищем самый длинный префикс из словаря, с которого начинается слово
        matched_value = None
        matched_length = 0
        
        for prefix, value in number_word_dict.items():
            if word.startswith(prefix) and len(prefix) > matched_length:
                matched_value = value
                matched_length = len(prefix)
        
        # Если ничего не нашли — пропускаем (по условию такого быть не должно)
        if matched_value is None:
            continue
        
        # Обрабатываем тысячи и миллионы
        if matched_value == 1000:  # "тысяча", "тысячи", "тысяч"
            current *= 1000
            total += current
            current = 0
        elif matched_value == 1000000:  # "миллион"
            current *= 1000000
            total += current
            current = 0
        else:
            # Обычное число — прибавляем к текущей сумме
            current += matched_value
    
    # Добавляем остаток (если после тысяч/миллионов были ещё числа)
    total += current
    
    return total


# Основной блок программы
try:
    # Ввод пользователя
    user_input = input("Введите число словами: ")
    
    # Преобразуем строку в число
    result = transform_string_to_integer(user_input)
    
    # Выводим результат
    print(result)

except Exception as e:
    print(f"Ошибка: {e}")
```

# Задание 5. Разложение факториала на простые множители

**Задача:** для числа `n` вывести разложение `n!` на простые множители.

**Ключевая идея:** не вычислять `n!` явно (для больших `n` это огромное число).
Вместо этого используем **формулу Лежандра**:
степень(p) = n//p + n//p² + n//p³ + ...


**Алгоритм:**
1. Перебираем все простые числа `p` от 2 до `n`.
2. Для каждого `p` считаем степень по формуле Лежандра.
3. Формируем строку: `p^k` (если `k > 1`) или `p` (если `k = 1`).
4. Соединяем через ` * `.

**Функции:**
- `is_prime(num)` — проверка на простоту.
- `prime_factors_of_factorial(n)` — словарь `{простое: степень}`.
- `decompose_factorial(n)` — формирует итоговую строку.

**Тесты:**
- `5` → `5! = 2^3 * 3 * 5` 
- `22` → `22! = 2^19 * 3^9 * 5^4 * 7^3 * 11^2 * 13 * 17 * 19` 
- `25` → `25! = 2^22 * 3^10 * 5^6 * 7^3 * 11^2 * 13 * 17 * 19 * 23` 

```python
def is_prime(num):
    """
    Проверяет, является ли число простым.
    
    Простое число делится только на 1 и на само себя.
    """
    if num < 2:
        return False
    # Проверяем делители от 2 до sqrt(num)
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True


def prime_factors_of_factorial(n):
    """
    Возвращает словарь {простое_число: степень} для разложения n! на простые множители.
    
    Использует формулу Лежандра: степень(p) = n//p + n//p^2 + n//p^3 + ...
    Это позволяет не вычислять сам факториал (который может быть огромным).
    """
    factors = {}
    
    # Перебираем все простые числа от 2 до n
    for p in range(2, n + 1):
        if not is_prime(p):
            continue
        
        # Формула Лежандра: считаем степень p в n!
        power = 0
        divisor = p
        while divisor <= n:
            power += n // divisor
            divisor *= p
        
        factors[p] = power
    
    return factors


def decompose_factorial(n):
    """
    Вычисляет n! и возвращает строку с разложением на простые множители.
    
    Формат: "n! = 2^k * 3^m * 5 * 7 * ..."
    Множители — по возрастанию. Степень указывается только если > 1.
    
    Параметры:
        n (int) — целое число.
    
    Возвращает:
        str — строка с разложением.
    """
    # Получаем словарь {простое: степень}
    factors = prime_factors_of_factorial(n)
    
    # Формируем строку разложения
    parts = []
    for prime in sorted(factors.keys()):  # по возрастанию
        power = factors[prime]
        if power == 1:
            parts.append(str(prime))         # без степени
        else:
            parts.append(f"{prime}^{power}") # со степенью
    
    # Соединяем через " * "
    decomposition = " * ".join(parts)
    
    return f"{n}! = {decomposition}"


# Основной блок программы
try:
    # Ввод пользователя
    n = int(input("Введите число n: "))
    
    # Проверка: n должно быть >= 0
    if n < 0:
        print("Некорректный ввод")
    else:
        # Получаем разложение и выводим
        result = decompose_factorial(n)
        print(result)

except ValueError:
    print("Некорректный ввод")
```

# Задание 6. Замена дублирующихся букв

**Задача:** последовательно заменять две одинаковые буквы на следующую по алфавиту.
Повторять, пока в строке не останутся только уникальные символы.

**Особенности:**
- Буквы **не обязаны быть смежными** — ищем **любые две одинаковые**.
- `zz` → `a` (циклическая замена).
- Порядок вывода — **любой**.

**Алгоритм (через частоты):**
1. Считаем частоты букв (`counts`).
2. Пока есть буква с частотой ≥ 2:
   - `pairs = counts[ch] // 2` — сколько пар заменяем.
   - `counts[ch] = counts[ch] % 2` — остаток (0 или 1).
   - Следующая буква: `alphabet[(idx + 1) % 26]` — с циклом `z → a`.
   - `counts[next_ch] += pairs`.
3. Собираем буквы с частотой > 0.

**Почему через частоты?**
- Порядок букв **не важен** (задание допускает любой порядок).
- Замена пар — **ассоциативна**: неважно, какие именно две буквы из трёх схлопнуть первыми.
- Каскад обрабатывается циклом `while changed`.

**Тесты:**
- `zzzab` → `zc` (или `cz`) 
- `hhakafh` → `hkfib` 

```python
from string import ascii_lowercase


def replace_duplicates(s):
    """
    Последовательно заменяет две одинаковые буквы на следующую по алфавиту.
    Операция повторяется, пока в строке не останутся только уникальные символы.
    
    Принцип:
    - Считаем частоты букв.
    - Если буква встречается >= 2 раз, каждые 2 её вхождения заменяем
      на одну следующую по алфавиту букву (z → a).
    - Повторяем, пока есть буквы с частотой >= 2.
    
    Параметры:
        s (str) — строка из латинских букв в нижнем регистре.
    
    Возвращает:
        str — строка с уникальными буквами (порядок любой).
    """
    # Алфавит для циклического перехода: z → a
    alphabet = ascii_lowercase  # 'abcdefghijklmnopqrstuvwxyz'
    
    # Считаем частоты букв
    counts = {}
    for ch in s:
        counts[ch] = counts.get(ch, 0) + 1
    
    # Повторяем, пока есть буквы с частотой >= 2
    changed = True
    while changed:
        changed = False
        
        # Проходим по всем буквам, у которых частота >= 2
        for ch in list(counts.keys()):
            if counts[ch] >= 2:
                # Сколько пар можно заменить
                pairs = counts[ch] // 2
                # Остаток (0 или 1)
                counts[ch] = counts[ch] % 2
                
                # Следующая буква по алфавиту (z → a)
                idx = alphabet.index(ch)
                next_ch = alphabet[(idx + 1) % 26]
                
                # Добавляем pairs новых букв
                counts[next_ch] = counts.get(next_ch, 0) + pairs
                
                changed = True
    
    # Собираем результат: только буквы с частотой 1
    result = ''.join(ch for ch, cnt in counts.items() if cnt > 0)
    
    return result


# Основной блок программы
try:
    # Ввод пользователя
    user_input = input("Введите строку: ").strip().lower()
    
    # Проверка: строка состоит только из латинских букв
    if not user_input.isalpha() or not user_input.isascii():
        print("Некорректный ввод")
    else:
        # Заменяем дубликаты
        result = replace_duplicates(user_input)
        print(result)

except Exception as e:
    print(f"Ошибка: {e}")
```
